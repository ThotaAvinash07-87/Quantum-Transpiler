```python
from qiskit import QuantumCircuit
import matplotlib.pyplot as plt

# 1. Initialize a 3-qubit quantum register
# (In reversible logic, inputs also act as control lines)
qc = QuantumCircuit(3)

# 2. Add a Toffoli gate (Controlled-Controlled-NOT)
# If qubit 0 AND qubit 1 are high, flip qubit 2. 
qc.ccx(0, 1, 2)

# 3. Render the circuit directly
qc.draw('mpl')
```
